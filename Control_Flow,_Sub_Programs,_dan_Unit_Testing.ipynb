## **Control Flow**

### **Conditional Statement**
* The ability of a program to alter its execution sequence is called branching.
* The if block will be executed only if the condition is True.
* Use elif & else statement to handle multiple conditions.

```python
nilai = 65

if nilai>=80:
   print("Selamat! Anda mendapat nilai A")
   print("Pertahankan!")
elif nilai>=70:
   print("Hore! Anda mendapat nilai B")
   print("Tingkatkan!")
elif nilai>=60:
   print("Hmm.. Anda mendapat nilai C")
   print("Ayo semangat!")
else:
   print("Waduh, Anda mendapat nilai D")
   print("Yuk belajar lebih giat lagi!")
```

### **Loops**




#### `while loops`

**while** loop instruct computer to continuously execute code based on the value of a condition.

```python
x = 7

while x > 0:
  print("positive x=" + str(x))
  x = x - 1

print("now x=" + str(x))
```

**Warning** to avoid infinity loop like this

```python
while x > 0:
  print("positive x=" + str(x))
  x = x + 1 # Wrong conditions
  print("now x=" + str(x))
```

#### `for loop`
**for** loop iterates over a sequence of values.

```python
for x in range(3):
  print("x=" + str(x))
```

**for** loop also can do this

```python
sentences = ['Dicoding', "Indonesia"]
for word in sentences:
  print(word)
```

### **Break & Continue**

Both **while** and **for** loops can be interrupted using the `break` keyword.

```python
for x in range(3):
    print("x=" + str(x))
    if x == 1:
        break  # quit from loop
```

Use the **continue** keyword to skip the current iteration and continue with the next one.

```python
for x in range(3, 0, -1):
    if x % 2 == 0:
        continue  # skip even
    print(x)
```

### **Exception Handling**

Python exception handling is a mechanism that allows you to manage runtime errors gracefully using try, except, else, finally, and raise keywords. This prevents a program from crashing abruptly and makes the code more robust.

**Core Keywords and Usage**
The primary structure for handling exceptions involves a try block and a corresponding except block.
*   `try`: Contains the code that might cause an error.
*   `except`: Catches and handles the exception if one occurs in the try block.
*   `else`: Executes only if the code in the try block runs without any exceptions.
*   `finally`: Executes regardless of whether an exception occurred or was handled; it is useful for cleanup operations like closing files.
*  raise: Used to forcefully trigger an exception manually when a specific condition is met.

#### **Example: Basic Exception Handling**

Here is a basic example demonstrating how to handle a `ZeroDivisionError`:

```python
try:
    numerator = 10
    denominator = 0
    result = numerator / denominator # This line will raise a ZeroDivisionError
except ZeroDivisionError:
    print("Error: Denominator cannot be 0.")
finally:
    print("Execution complete.")
```

In this example, the division by zero in the try block causes an exception. The program immediately jumps to the except `ZeroDivisionError` block, prints the error message, and then the finally block runs.

#### **Catching Specific vs. Multiple Exceptions**
It is best practice to catch specific exceptions to avoid masking unexpected errors.

**Catching specific exceptions**

You can use multiple except blocks for different error types.

```python
try:
    x = int("str") # This causes a ValueError
    inv = 1 / x
except ValueError:
    print("Not a valid number!")
except ZeroDivisionError:
    print("Zero has no inverse!")
```

**Catching multiple exceptions with the same handler**

You can group exceptions in a tuple.

```python
try:
    # Code that might raise ValueError or TypeError
    pass
except (ValueError, TypeError) as e:
    print(f"An error occurred: {e}")
```

#### **Raising Custom Exceptions**

You can define your own exceptions by creating a new class that inherits from the built-in Exception class.

```python
class InvalidAgeError(Exception):
    """Raised when an age is less than zero."""
    pass

def check_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative.")
    print("Access granted.")

try:
    check_age(-5)
except InvalidAgeError as e:
    print(e)
```

## **Unit Testing**

Unit testing is a software testing process that ensures every unit/function of a program is tested. If the functionality of the application we're building consists of procedures and functions that we write, we need to conduct unit tests for each procedure or function.

We'll learn unit testing using one of Python's built-in libraries, unittest. Like a testing framework, the unittest library supports the following essential features:

* Automated testing.
* Reusable code for the start (setup) and end (shutdown) of a process.
* Unifying multiple tests into a single collection.
* Separation of the testing framework from the reporting framework.

The unittest library supports several important object-oriented concepts, including the following.
* A test fixture represents the preparation required to perform one or more tests and the cleanup process. Some examples include setting up a test database, a test directory, or starting a server process.
* A test case is a unit of testing that checks a set of responses to a subset of input. The unittest library provides base classes and TestCase that can be used to create new test cases.
* A test suite is a collection of test cases, a collection of test suites, or a combination of both. It is useful for grouping tests to be executed together.
* A test runner is a component that orchestrates the execution of tests and provides output to the user. In this case, the runner can use graphical displays, textual displays, or return special values ​​indicating the test results.

```python
# Fungsi yang akan diuji
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b
```

```python
import unittest

class TestMathOperations(unittest.TestCase):
    # Test Case 1
    def test_add(self):
        self.assertEqual(add(10, 5), 15)
        self.assertEqual(add(-1, 1), 0)
        self.assertEqual(add(-1, -1), -2)

    # Test Case 2
    def test_subtract(self):
        self.assertEqual(subtract(10, 5), 5)
        self.assertEqual(subtract(-1, 1), -2)
        self.assertEqual(subtract(-1, -1), 0)

    # Test Case 3
    def test_multiply(self):
        self.assertEqual(multiply(10, 5), 50)
        self.assertEqual(multiply(-1, 1), -1)
        self.assertEqual(multiply(-1, -1), 1)

    # Test Case 4
    def test_divide(self):
        self.assertEqual(divide(10, 5), 2)
        self.assertEqual(divide(-1, 1), -1)
        self.assertEqual(divide(-1, -1), 1)
        with self.assertRaises(ValueError):
            divide(10, 0)

if __name__ == '__main__':
    # Test Runner
    unittest.main(argv=[''], verbosity=2, exit=False)
```
